# 01 — Python Memory System

## 1. Variables as Names

### Anchor

A variable is not a box.

A variable is a **name tag pointing to an object**.

Python memory can be imagined as:

name → object

Example:

x → 10

The important question is not “what is inside x?”  
The better question is:

**What object does the name x point to?**

```python
x = 10
y = x
x = 20

print(y)
```

### Result

`y` prints `10`.

Reason:

`y = x` does not create a living relationship where `y` follows `x`.

It only makes `y` point to the same object that `x` pointed to at that moment.

When `x = 20`, only the name `x` moves.

Assignment is not reactive.

it computes the expression immediately using the current value.

It does not create a live formula like Excel.

```python
temperature = "cold"
state = temperature
temperature = "hot"

print(state)
```

```python
a = 3
b = a + 2
a = 10

print(b)
```

```python
score = 0
score = score + 1
score = score + 1

print(score)
```

### Reassignment as state update

`score = score + 1` means:

1. Read the current value of `score`
2. Create the new value
3. Move the name `score` to that new value

This is not a live relationship.

It is repeated updating of the same name.

A variable can represent the current state of a system.

Each reassignment updates that state step by step.


```python
energy = 10
energy = energy - 3
energy = energy * 2

print(energy)
```

```python
x = "5"
y = "2"

print(x + y)
```

### Types define meaning

The symbol `+` does different things depending on type.

For numbers:
5 + 2 → arithmetic addition

For strings:
"5" + "2" → text concatenation

Python interprets operations through type.

```python
x = "5"
y = 2

print(x + y)
```

### Type mismatch

Python does not guess my intention.

If I combine incompatible types, Python raises an error instead of silently converting.

"5" + 2 fails because one side is text and the other is a number.

```python
x = "5"
y = 2

print(int(x) + y)
```

```python
name = "Alessa"
score = 10
score = score + 5

print(f"{name} scored {score} points")
```

### F-strings

An f-string is text with live Python slots.

The `f` before the string tells Python:

Evaluate anything inside `{ }` and insert the result into the text.

Example:

name = "Alessa"
score = 15

f"{name} scored {score} points"

becomes:

"Alessa scored 15 points"

The arithmetic happens before the f-string displays the value.

Yes, exactly.

You use an f-string because you want to mix:

fixed text: "The result is"
live Python value: x + y

Without f-string, this fails

because Python cannot concatenate:

string + integer

```python
x = 4
y = 3

print(f"The result is {x + y}")
```

You could manually convert:

print("The result is " + str(x + y))

But f-string is cleaner:

print(f"The result is {x + y}")

The {x + y} means:

calculate this Python expression first, then insert the result into the text.

```python
age = 25
next_age = age + 1

print(f"Next year I will be {next_age}")
```

## Final Concept — Expression vs Assignment

An expression produces a value.

Example:

3 + 2

produces 5, but Python does not remember it unless I assign it.

Assignment gives a produced value a name.

result = 3 + 2

Now the name `result` points to the object/value 5.

Core distinction:

- expression → produces
- assignment → remembers

```python
result = 3 + 2
result = result * 10

print(f"The final result is {result}")
```

# Session 1 Summary — Python Memory System

Today I learned that Python memory is based on:

name → object

## Main anchors

1. Variables are name tags, not boxes.
2. Assignment is not a live relationship.
3. Reassignment moves a name to a new object/value.
4. Types define how Python interprets operations.
5. F-strings insert live values into text.
6. Expressions produce values.
7. Assignment remembers values.

## Core failure modes

- Thinking one variable follows another automatically.
- Thinking `"5"` and `5` are the same.
- Forgetting that Python follows type rules.
- Confusing producing a value with storing a value.

## Diagnostic questions

When I see Python code, I should ask:

1. What names exist?
2. What object/value does each name point to?
3. What type is each value?
4. Am I producing a value or storing it?
5. Did I expect a live relationship where Python only made an assignment?
