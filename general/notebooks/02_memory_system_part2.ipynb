## Shared State, Mutation, and References

In Part 1, I learned:

name → object

Now I add a deeper idea:

multiple names can point to the same mutable object.

Core anchor:

- Reassignment moves a name.
- Mutation changes the object itself.
- Shared references mean many names can see the same mutation.

## 1. Mutation

Mutation means changing an object internally.

A list is mutable: I can modify the same list object without moving the name.

```python
a = [1, 2]
b = a

a.append(3)

print(b)
```

```python
### Result

`b` prints `[1, 2, 3]`.

Reason:

`b = a` makes both names point to the same list.

`a.append(3)` does not move `a` to a new object.  
It changes the existing list internally.

Because `a` and `b` share the same object, both see the mutation.
```

```python
a = [1, 2]
b = a

a = [1, 2, 3]

print(b)
```

is reassignment.

It creates a new list and moves only the name a.

b stays pointing to the old list

```python
a = [1, 2]
b = a

a = [9, 9]
a.append(3)

print(a)
print(b)
```

### Mutation after reassignment

Reassignment can break sharing.

After `a = [9, 9]`, `a` and `b` no longer point to the same list.

So `a.append(3)` only affects `a`'s new list.

```python
def add_item(items):
    items.append("new")

x = ["old"]
add_item(x)

print(x)
```

### Function side effects

Function parameters are local names.

When I pass a mutable object like a list into a function, the function parameter can point to the same object.

If the function mutates that object, the outside state changes too.

This is called a side effect.

```python
def replace_items(items):
    items = ["new"]

x = ["old"]
replace_items(x)

print(x)
```

### Reassignment inside a function

Inside a function, the parameter name is local.

`items = ["new"]` moves only the local name `items`.

It does not mutate the original object passed from outside.

So `x` stays as `["old"]`.

```python
def modify(items):
    items.append("new")
    items = ["reset"]

x = ["old"]
modify(x)

print(x)
```

A parameter is a temporary local nickname.
Mutation through the nickname changes the shared object.
Reassignment of the nickname only moves the nickname.

The parentheses () mean:

“call / execute this function now.”

This:

def modify(items):

defines the function.

This:

modify(x)

runs the function.

```python
def greet(name):
    print(f"Hello {name}")

greet("Alessa")
```

```python
def clean_list(items):
    items = ["clean"]
    return items

x = ["old"]
result = clean_list(x)

print(x)
print(result)
```

x still points to ["old"]
items = ["clean"] only moves the local name items
return items sends ["clean"] back
result receives that returned list

So the function did not modify x. It created and returned a new object.

```python
def clean_list(items):
    items = ["clean"]
    return items

x = ["old"]

x = clean_list(x)

print(x)
```

To make x itself become ["clean"], you need to return the new list and reassign x outside.

If the function mutates, x changes through shared object.
If the function reassigns locally, x changes only if you return and reassign outside.

```python
def modify(items):
    items.append("new")
    items = ["reset"]
    return items

x = ["old"]

result = modify(x)

print(x)
print(result)
```

def modify(items):
    items.append("new")   # mutation
    items = ["reset"]     # local reassignment
    return items          # returns ["reset"]

x = ["old"]
result = modify(x)

What happens:

items and x point to the same list: ["old"]
items.append("new") mutates that shared list → x becomes ["old", "new"]
items = ["reset"] moves only the local name items to a new list
return items returns the new list ["reset"]
result receives ["reset"]

Mutation changed x.
Reassignment created the returned result.

# Session 2 Summary — Shared State

## Core distinctions

- Mutation changes an object internally.
- Reassignment moves a name.
- Multiple names can point to the same object.
- Function parameters are local names.
- Side effects happen when shared objects are mutated.
- `return` sends a value/object back outside the function.
- Outside reassignment is needed to replace the outside variable.

## Core diagnostic question

Am I:
1. mutating an object,
2. moving a name,
3. or returning a new object?
