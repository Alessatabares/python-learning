# 03 — Decision System Part 1

## 1. Truth and Gates

### Anchor

A condition is a gate.

Python asks:

Is this condition True or False?

If True → enter the gate.  
If False → skip the gate.

A program decision is not emotion or intuition.  
It is a gate controlled by a Boolean value.


"True"   # text
True     # Boolean value

```python
is_adult = True

print(is_adult)
print(type(is_adult))
```

## 2. Comparisons

Comparisons ask questions.

Examples:

- Is 5 greater than 3?
- Is 10 equal to 2?
- Is one value different from another?

Comparisons produce Boolean values:

True or False.

```python
print(5 > 3)
print(10 == 2)
print(4 != 1)
```

>   # greater than
==  # equal to
!=  # not equal to

### Assignment vs comparison

`=` assigns a value to a name.

`==` compares two values and returns True or False.

```python
age = 18

print(age == 18)
print(age > 18)
print(age < 18)
```

```python
probability = 0.72
threshold = 0.50

print(probability > threshold)
print(probability == threshold)
print(probability < threshold)
```

A threshold is a boundary.

If a value is above the threshold, Python can route the system one way.

If it is below, Python can route it another way.

```python
probability = 0.72
threshold = 0.50

if probability > threshold:
    print("Predict class 1")
```

### If gate and indentation

An `if` statement routes the program into a path only when the condition is True.

Indentation shows which code belongs inside that path.

Indentation means:

“This line belongs inside the condition.”

In machine learning, a threshold can decide a class:

- class 1 = positive category
- class 0 = negative category

Example:

if probability > threshold:
    predict class 1

```python
probability = 0.32
threshold = 0.50

if probability > threshold:
    print("Predict class 1")
else:
    print("Predict class 0")
```

### Else path

`else` is the fallback path.

If the `if` condition is False, Python enters `else`.

In threshold logic:

- probability > threshold → class 1
- otherwise → class 0

```python
probability = 0.50
threshold = 0.50

if probability > threshold:
    print("Predict class 1")
else:
    print("Predict class 0")
```

```python
### Boundary case

A boundary is the exact cutoff point where a decision changes.

With:

probability > threshold

the value must be strictly above the threshold.

So:

0.50 > 0.50

is False.

If I want equality to count as positive, I need:

probability >= threshold
```

```python
probability = 0.50
threshold = 0.50

if probability >= threshold:
    print("Predict class 1")
else:
    print("Predict class 0")
```

```python
probability = 0.50
threshold = 0.50
patient_has_symptoms = True

if probability >= threshold:
    print("Model says: class 1")

if patient_has_symptoms:
    print("Clinical context supports concern")

print("Decision review complete")
```

# Session Summary — Decision System Part 1

## Core anchor

A condition is a gate.

Python asks whether a condition is True or False.

If True, Python enters the gate.
If False, Python skips it or goes to `else`.

## Main ideas

- `True` and `False` are Boolean values.
- Comparisons create Boolean values.
- `if` creates a decision gate.
- `else` creates a fallback path.
- Indentation shows what belongs inside the gate.
- A threshold is a decision boundary.
- `>` means strictly greater.
- `>=` means greater than or equal.

## Failure modes

- Confusing `=` assignment with `==` comparison.
- Forgetting that indentation defines structure.
- Thinking equality counts when using `>`.
- Assuming Python understands intention instead of exact conditions.

## Diagnostic question

Which condition is Python checking, and which path does that condition open?
