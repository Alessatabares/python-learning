# 04 — Decision System Part 2

## Multiple Gates vs Exclusive Branches

### Anchor

In Part 1, I learned that an `if` statement is a gate.

Now I learn two different decision structures:

1. Multiple independent gates
2. One exclusive decision tree

Independent gates can all open.

An exclusive decision tree chooses only one path.

```python
fever = True
cough = True

if fever:
    print("Check temperature pattern")

if cough:
    print("Check respiratory symptoms")

print("Initial screening complete")
```

### Multiple independent gates

Separate `if` statements are independent.

Python checks each one.

If multiple conditions are True, multiple blocks can run.

```python
fever = True
cough = False

if fever:
    print("Check temperature pattern")

if cough:
    print("Check respiratory symptoms")

print("Initial screening complete")
```

```python
score = 85

if score >= 60:
    print("Pass")

if score >= 80:
    print("Strong performance")

if score >= 95:
    print("Excellent")
```

```python
score = 85

if score >= 95:
    print("Excellent")
elif score >= 80:
    print("Strong performance")
elif score >= 60:
    print("Pass")
else:
    print("Fail")
```

if score >= 95:      # False
elif score >= 80:    # True → enter here and stop
elif score >= 60:    # not checked

Anchor:

Multiple if = many doors can open.
if/elif/else = one ranked pathway is selected.

```python
score = 98

if score >= 60:
    print("Pass")

if score >= 80:
    print("Strong performance")

if score >= 95:
    print("Excellent")
```

```python
score = 98

if score >= 95:
    print("Excellent")
elif score >= 80:
    print("Strong performance")
elif score >= 60:
    print("Pass")
else:
    print("Fail")
```

elif = “alternative if, but only if previous conditions failed”
else = “fallback if none of the above worked”

```python
score = 85
late_submission = True

if score >= 95:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

if late_submission:
    print("Warning: late submission")

print(f"Final grade: {grade}")
```

The decision tree chooses/stores the result.
Printing happens only where print() appears.

# Decision System Part 2 — Summary

## Core anchor

Python decisions can work in two major ways:

1. Multiple independent gates
2. One exclusive decision tree

## Independent gates

Separate `if` statements are checked independently.

If several conditions are True, several blocks can run.

Example:

if fever:
    print("Check temperature")

if cough:
    print("Check respiratory symptoms")

Both can print if both are True.

## Exclusive decision tree

`if / elif / else` chooses one pathway.

Python checks from top to bottom.

Once one condition is True, Python enters that branch and skips the rest.

Example:

if score >= 95:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

Only one grade is assigned.

## Else

`else` is the fallback path.

It runs only if none of the previous conditions were True.

## Important distinction

A decision block can assign a value without printing anything.

Printing only happens where `print()` appears.

Example:

grade = "B"

does not show anything.

print(f"Final grade: {grade}")

shows the result.

## Failure modes

- Using many `if` when only one branch should happen.
- Using `elif` when multiple independent signals should be checked.
- Thinking assignment prints automatically.
- Forgetting that Python checks `if / elif / else` from top to bottom.
- Not knowing which branch owns which consequence.

## Final anchor

Multiple `if` statements = many doors can open.

`if / elif / else` = one ranked pathway is selected.

## Diagnostic question

Am I checking:

1. independent signals that can all be true?

or

2. mutually exclusive options where only one path should win?

If several things can happen together → use multiple `if`.

If only one category/result should be chosen → use `if / elif / else`.

Am I opening many doors, or choosing one path?
