## While Loops

### Anchor

A `while` loop is a monitor.

Python keeps repeating an action while a condition remains True.

Structure:

while condition:
    action
    update state

Core danger:

If the condition never becomes False, the loop never stops.

```python
count = 0

while count < 3:
    print(count)
    count = count + 1
```

It prints the current value of count at that moment, before the update.

cycle 1: print count → 0, then update to 1
cycle 2: print count → 1, then update to 2
cycle 3: print count → 2, then update to 3
stop: 3 < 3 is False

```python
count = 0

while count < 3:
    count = count + 1
    print(count)
```

cycle 1: count 0 → 1, print 1
cycle 2: count 1 → 2, print 2
cycle 3: count 2 → 3, print 3
stop: 3 < 3 is False

Anchor:

In a while loop, order inside the loop body matters.
Print before update → see old state.
Update before print → see new state.

```python
count = 0

while count < 3:
    print(count)
```

Here count never changes.

So Python keeps checking:

0 < 3

So it prints forever:

0
0
0
0
...

Anchor:

A while loop needs a changing state.
If the state never changes, the condition may never become False.

A while loop needs two things:

A condition to monitor.
A state update that eventually makes the condition false.

Without the update, the loop can run forever.

```python
energy = 5

while energy > 0:
    print(f"Energy: {energy}")
    energy = energy - 2

print("Done")
```

energy = 5 → print 5 → energy becomes 3
energy = 3 → print 3 → energy becomes 1
energy = 1 → print 1 → energy becomes -1
energy = -1 → condition energy > 0 is False → stop loop
print Done

Condition monitors: energy > 0
It stops when energy is no longer greater than 0.

Anchor:

Indentation decides what belongs inside the loop.
Done is not indented, so it happens after the loop.

```python
energy = 10
steps = 0

while energy > 0:
    if energy >= 5:
        print(f"High energy: {energy}")
    else:
        print(f"Low energy: {energy}")

    energy = energy - 3
    steps = steps + 1

print(f"Stopped after {steps} steps")
print(f"Final energy: {energy}")
```

cycle 1: energy 10 → subtract 3 → 7
cycle 2: energy 7  → subtract 3 → 4
cycle 3: energy 4  → subtract 3 → 1
cycle 4: energy 1  → subtract 3 → -2
stop: -2 > 0 is False

Key anchor:

The loop prints the state before the update, but the final value is after the last update.

```python
energy = 12
steps = 0
warning_count = 0

while energy > 0:
    if energy <= 5:
        warning_count = warning_count + 1
        print(f"Warning energy: {energy}")
    else:
        print(f"Energy ok: {energy}")

    energy = energy - 4
    steps = steps + 1

print(f"Total steps: {steps}")
print(f"Warnings: {warning_count}")
print(f"Final energy: {energy}")
```

start energy = 12

cycle 1: print 12 → energy = 8 → steps = 1
cycle 2: print 8  → energy = 4 → steps = 2
cycle 3: print 4  → energy = 0 → steps = 3

check again: energy > 0
0 > 0 is False
stop

The 4th check happens, but it is not a cycle because the loop body does not run.

Anchor:

A cycle only counts when Python enters the loop body.
A failed condition check stops the loop but does not count as a cycle.

# Session Summary — Repetition System Part 2

## Core anchor

A `while` loop is a monitor.

Python keeps repeating while a condition remains True.

Structure:

while condition:
    action
    update state

## Main ideas

- `while` loops repeat based on a condition.
- The condition is checked before every cycle.
- A cycle only happens if the condition is True.
- The loop must update state so the condition can eventually become False.
- If state never changes, the loop can become infinite.
- Order matters: printing before update shows old state; updating before print shows new state.
- Indentation decides what is inside the loop and what happens after the loop.

## Core failure modes

- Forgetting to update the variable being monitored.
- Creating an infinite loop.
- Thinking the final failed check counts as a cycle.
- Confusing printed state with final state.
- Misreading indentation and thinking code is inside the loop when it is outside.

## Diagnostic questions

1. What condition is being monitored?
2. What variable/state changes each cycle?
3. Does the update make the condition eventually become False?
4. Which lines are inside the loop?
5. Which lines happen after the loop?
6. Am I printing the state before or after the update?
7. Did Python enter the loop body, or only check the condition and stop?

## Final anchor

A `for` loop walks through a known collection.

A `while` loop watches a condition.

`for` = conveyor belt  
`while` = monitor
