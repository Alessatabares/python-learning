Quieres un programa muy simplificado que pregunte:

Peso del paciente en kg.
Dosis indicada en mg/kg.

Y calcule:

dosis total = peso × dosis por kg

```python
peso = float(input("peso en kg:"))
dosis = 5
dosis_total = peso * dosis
print(f"Dosis total: {dosis_total} mg")
```

Quieres un programa que pregunte al usuario:

distancia recorrida en kilómetros
tiempo empleado en minutos

Y calcule:

ritmo = tiempo / distancia

```python
distancia = int(input("cual fue la distancia recorrida en km"))
tiempo = int(input("tiempo en minutos"))
ritmo = tiempo / distancia
print(f" tu ritmo fue de {ritmo} minutos por km")
```
